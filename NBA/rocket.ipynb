## BMI 身體質量指數計算

**BMI公式：** BMI = 體重(kg) / 身高(m)²

**BMI分類標準：**
- BMI < 18.5：體重過輕
- 18.5 ≤ BMI < 24：正常範圍
- 24 ≤ BMI < 27：體重過重
- BMI ≥ 27：肥胖

---


### 題目一：基礎BMI計算

**題目：**
1. 請輸入身高（單位：公分）和體重（單位：公斤）
2. 計算BMI值並顯示結果（保留小數點後2位）

**提示：**
- 身高需要從公分轉換為公尺（除以100）
- 使用 BMI = 體重 / 身高² 的公式
- 使用 f-string 格式化輸出


```python
# 題目一：基礎BMI計算（請在這裡寫你的程式碼）
# 範例：身高170公分，體重65公斤

# 方法1：使用 input() 輸入
# height_cm = float(input("請輸入身高（公分）："))
# weight = float(input("請輸入體重（公斤）："))

# 方法2：直接給定數值（練習時使用）
height_cm = 170
weight = 65

# 將公分轉換為公尺
height_m = height_cm / 100

# 計算BMI
bmi = weight / (height_m ** 2)

# 顯示結果（保留小數點後2位）
print(f"身高：{height_cm} 公分")
print(f"體重：{weight} 公斤")
print(f"BMI：{bmi:.2f}")
```

### 題目二：BMI判斷與分類

**題目：**
1. 計算BMI值
2. 根據BMI值判斷體重狀態（過輕、正常、過重、肥胖）
3. 顯示BMI值和體重狀態

**提示：**
- 使用 if-elif-else 進行條件判斷
- 參考上面的BMI分類標準


```python
# 題目二：BMI判斷與分類（請在這裡寫你的程式碼）

# 輸入資料
height_cm = 175
weight = 70

# 計算BMI
height_m = height_cm / 100
bmi = weight / (height_m ** 2)

# 判斷體重狀態
if bmi < 18.5:
    status = "體重過輕"
elif bmi < 24:
    status = "正常範圍"
elif bmi < 27:
    status = "體重過重"
else:
    status = "肥胖"

# 顯示結果
print(f"身高：{height_cm} 公分")
print(f"體重：{weight} 公斤")
print(f"BMI：{bmi:.2f}")
print(f"體重狀態：{status}")
```

### 題目三：完整的BMI計算器（進階題）

**題目：**
1. 使用 input() 讓使用者輸入身高和體重
2. 計算BMI值
3. 判斷體重狀態並顯示詳細資訊
4. 加入錯誤處理（檢查輸入是否為正數）

**挑戰：**
- 可以加入迴圈，讓程式可以連續計算多個人的BMI
- 檢查輸入值的合理性（例如：身高不超過250公分，體重不超過300公斤）


```python
# 題目三：完整的BMI計算器（請在這裡寫你的程式碼）

# 輸入資料
height_cm = float(input("請輸入身高（公分）："))
weight = float(input("請輸入體重（公斤）："))

# 檢查輸入是否合理
if height_cm <= 0 or weight <= 0:
    print("錯誤！身高和體重必須大於0")
elif height_cm > 250:
    print("警告！身高輸入可能有誤（超過250公分）")
elif weight > 300:
    print("警告！體重輸入可能有誤（超過300公斤）")
else:
    # 計算BMI
    height_m = height_cm / 100
    bmi = weight / (height_m ** 2)
    
    # 判斷體重狀態
    if bmi < 18.5:
        status = "體重過輕"
        suggestion = "建議增加營養攝取"
    elif bmi < 24:
        status = "正常範圍"
        suggestion = "保持良好習慣"
    elif bmi < 27:
        status = "體重過重"
        suggestion = "建議適度運動與飲食控制"
    else:
        status = "肥胖"
        suggestion = "建議諮詢專業醫師或營養師"
    
    # 顯示詳細結果
    print("\n" + "="*30)
    print("BMI 計算結果")
    print("="*30)
    print(f"身高：{height_cm} 公分 ({height_m:.2f} 公尺)")
    print(f"體重：{weight} 公斤")
    print(f"BMI：{bmi:.2f}")
    print(f"體重狀態：{status}")
    print(f"建議：{suggestion}")
    print("="*30)
```

### 練習題：自己試試看

**練習1：** 計算以下人員的BMI
- 小明：身高160公分，體重50公斤
- 小華：身高180公分，體重80公斤
- 小美：身高165公分，體重45公斤

**練習2：** 修改程式，讓它可以計算多個人的BMI（使用迴圈）

**練習3：** 加入功能，計算理想體重範圍
- 理想BMI範圍：18.5 ~ 24
- 理想體重 = 理想BMI × 身高²


```python
# 練習題：計算多個人的BMI
# 請在這裡寫你的程式碼

people = [
    {"name": "小明", "height": 160, "weight": 50},
    {"name": "小華", "height": 180, "weight": 80},
    {"name": "小美", "height": 165, "weight": 45}
]

print("="*40)
print("BMI 計算結果列表")
print("="*40)

for person in people:
    name = person["name"]
    height_cm = person["height"]
    weight = person["weight"]
    
    # 計算BMI
    height_m = height_cm / 100
    bmi = weight / (height_m ** 2)
    
    # 判斷狀態
    if bmi < 18.5:
        status = "體重過輕"
    elif bmi < 24:
        status = "正常範圍"
    elif bmi < 27:
        status = "體重過重"
    else:
        status = "肥胖"
    
    # 顯示結果
    print(f"\n{name}：")
    print(f"  身高：{height_cm} 公分")
    print(f"  體重：{weight} 公斤")
    print(f"  BMI：{bmi:.2f}")
    print(f"  狀態：{status}")

print("\n" + "="*40)
```
