```python
chinese=int(input("國文:"))
english=int(input("英文:"))
math=int(input("數學:"))

total=chinese + english + math
print("總分:",total,"分",sep="")
average = round(total/3, ndigits=4)
print("平均:",average,"分",sep="")
print("平均:"+str(average)+"分",sep="")

```
